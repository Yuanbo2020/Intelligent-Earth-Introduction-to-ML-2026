## VAE Theory

The goal of a VAE is to (implicitly) learn a distribution $p(x),\,x \in X$, with $X$ a vector space, which is as similar as possible to an unknown target distribution $p^{*}(x)$, given a set of samples from this distribution $\mathcal{D}=\{x_1, ..., x_{n}\}$. VAEs make the assumption that $x$ depends on some set of underlying latent variables $z \in Z$, which are distributed according to some known distribution $p(z)$, and model the likelihood $p(x\mid z)$ and an approximation of the posterior distribution $p(z \mid x)$.

To help them learn, VAEs use variational inference. In this approach, instead of computing the posterior $p(z \mid x)$, which is often intractable, one searches for a more tractable distribution $q(z \mid x)$ which approximates the posterior. VAEs assume parameterised models of the likelihood and the approximate posterior, given by $p(x \mid z,\theta)$ and $q(z \mid x, \theta)$, where $\theta$ represent model parameters. To learn  $q(z \mid x, \theta)$, we maximise the Evidence Lower Bound (or ELBO) over the training data, which is given by

$$
ELBO = \sum_{i=1}^{n} \mathbb{E}_{q(z_{i} \mid x_{i},\theta)} \left[\log p(x_{i} \mid z_{i},\theta) \right] - D_{\mathrm{KL}}(q(z_{i}\mid x_{i},\theta) \, \| \, p(z_{i}))~,
$$
assuming the training samples are independently and identically distributed. 

To be able to compute the ELBO, we assume a Gaussian VAE:
$$
\begin{aligned}
p(x \mid z,\theta)&=\mathcal{N}(x; f(z; \theta_{f}), \sigma^{2} I)
\\

p(z)&=\mathcal{N}(z; 0, I)
\\

q(z\mid x,\theta)&=\mathcal{N}(z; g(x;\theta_{g}), \mathrm{diag}(h(x;\theta_{h})^{2}))~,
\end{aligned}
$$
where $f$, $g$ and $h$ are all deep neural networks parameterised by $\theta=\{\theta_{f},\theta_{g},\theta_{h}\}$, $\sigma$ is a hyperparameter. Our optimal network parameters are then given by

$$
\hat \theta_{f}, \hat \theta_{g}, \hat \theta_{h} = argmin_{\theta_{f}, \theta_{g}, \theta_{h}}  \left\{ \sum_{i=1}^{n} \mathbb{E}_{q(z_{i} \mid x_{i}, \theta_{g},\theta_{h})} \left[{\| x_{i}-f(z_{i};\theta_{f}) \|^{2} \over 2 \sigma^{2}}\right] + D_{\mathrm{KL}}(q(z_{i}\mid x_{i},\theta_{g},\theta_{h}) \, \| \, p(z_{i})) \right\}~.
$$

The KL term above can easily be computed analytically as it involves two normal distributions. When calculating the expectation over $q(z\mid x,\theta)$, which is usually carried out via sampling, a "reparameterisation trick" is used. One notes that sampling $z \sim q(z\mid x,\theta)$ is equivalent to sampling $\epsilon \sim \mathcal{N}(\epsilon;0,I)$ and computing $z=h(x;\theta_{h})\epsilon + g(x;\theta_{g})$, which allows gradient descent methods to be used to find $\hat \theta$.

> **Extension task**: show that the KL divergence term in this case can be analytically derived as: $D_{\mathrm{KL}}(q(z_{i}\mid x_{i},\theta_{g},\theta_{h}) \, \| \, p(z_{i}))= -{1\over 2} \sum^{J}_{j=1} \left[ 1 + \log(h_{j}^{2}) - h_{j}^{2} - g_{j}^{2} \right]$ where $J$ is the dimension of the latent vector.

For more on VAE theory, see: https://arxiv.org/abs/1606.05908 and https://arxiv.org/abs/1907.08956.

**solution — KL derivation (the theory extension).** For a diagonal Gaussian $q(z\mid x)=\mathcal N(\mu,\mathrm{diag}(\sigma^2))$ and standard-normal prior $p(z)=\mathcal N(0,I)$, the Gaussian KL formula gives

$$
D_{\mathrm{KL}}(q\|p)=\frac12\sum_{j=1}^{J}(\mu_j^2+\sigma_j^2-1-\log\sigma_j^2)
= -\frac12\sum_{j=1}^J(1+\log\sigma_j^2-\sigma_j^2-\mu_j^2).
$$

The first term of the negative ELBO encourages faithful reconstruction; the KL term keeps the approximate posterior close to the standard normal prior. Reparameterisation writes $z=\mu+\sigma\odot\epsilon$ with $\epsilon\sim\mathcal N(0,I)$ so gradients can flow through the sample. The reconstruction used below is a squared-error term and therefore corresponds to the Gaussian-likelihood formulation (up to choice of scale).

## Define network

Next we define the VAE.

### OPTIONAL EXERCISE

**Reparameterise and implement `forward()`.** Follow the diagram: image → encoder → `(mu, logvar)` → `z` → decoder → reconstruction. Hints: `torch.exp(0.5 * logvar)` gives standard deviation; `torch.randn_like` gives normal noise.

**Task 1:** complete the PyTorch VAE model below by implementing the `reparameterize` method which samples from the approximate posterior using the reparameterisation trick, and implementing the model's `forward` method, which returns the estimated parameters $f$,$g$ and $h$ of the likelihood and approximate posterior given a sample $x$.

```python
import torch.nn as nn
import torch.nn.functional as F

class VAE(nn.Module):
    def __init__(self):
        super(VAE, self).__init__()

        self.fc1 = nn.Linear(784, 400)
        self.fc21 = nn.Linear(400, 20)
        self.fc22 = nn.Linear(400, 20)
        self.fc3 = nn.Linear(20, 400)
        self.fc4 = nn.Linear(400, 784)

    def encode(self, x):
        h1 = F.relu(self.fc1(x))
        return self.fc21(h1), self.fc22(h1)# estimates mu, log-variance of approximate posterior q(z|x)

    def reparameterize(self, mu, logvar):
        
        
        std = torch.exp(0.5*logvar)
        eps = torch.randn_like(std)
        return mu + eps*std

    def decode(self, z):
        h3 = F.relu(self.fc3(z))
        return torch.sigmoid(self.fc4(h3))# estimates mu of likelihood p(x|z)

    def forward(self, x):
        
        # 1) mu,log-variance of the approximate posterior given x
        # 2) mu of the likelihood given a random sample from the posterior given x
        
        mu, logvar = self.encode(x.view(-1, 784))
        z = self.reparameterize(mu, logvar)
        recon = self.decode(z)
        
        return recon, mu, logvar# should return  (likelihood mu, posterior mu, posterior log-variance)
    
model = VAE()
```

**solution**

Compute $\sigma=\exp(\tfrac12\mathrm{logvar})$, sample $\epsilon\sim\mathcal N(0,I)$, and use $z=\mu+\epsilon\odot\sigma$. The encoder returns `(mu, logvar)`, the decoder maps sampled `z` to a reconstructed 784-vector, and `forward` returns `(recon, mu, logvar)`.

## Loss function

Next, we define the loss function to train the VAE model.

### OPTIONAL EXERCISE

**Negative ELBO.** The theory gives a reconstruction term plus Gaussian KL penalty; `x.view(-1, 784)` reshapes a batch of images into vectors.

**Task 2:** define a loss function to train the VAE model, which minimises the expression for the negative ELBO shown above.

```python
def loss_function(recon, x, mu, logvar):
    
    
    # The reconstruction term can be approximated with a single sample from the approximate posterior.
    # Use the analytical exppression for the KL term.
    # recon, mu and logvar are the outputs from model.forward(x).
    
    L2 = torch.sum((recon - x.view(-1, 784))**2)
    
    KLD = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())

    return L2 + 1.0 * KLD
```

**solution**

The code computes $\\sum(x-\\hat x)^2$ and $-\\tfrac12\\sum(1+\\mathrm{logvar}-\\mu^2-\\exp(\\mathrm{logvar}))$. The first is the scaled Gaussian reconstruction error, and the second is KL to a unit Gaussian. The exercise uses a unit reconstruction weight; a different observation noise variance changes the coefficient.

## Train

Finally, we train the VAE model.

```python
from torch import optim

optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)# this automatically batches up examples, adding a batch dimension
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=True)
```

### OPTIONAL EXERCISE

**Training and evaluation loops.** This is a longer task; a GPU is helpful but not required. Only run the following full-epoch loop after implementing the earlier optional model and loss.

**Task 3:** write some code to train the VAE using the training data and the loss function defined above.

```python
def train(epoch):
    model.train()
    
    train_loss = 0
    for batch_idx, (data, _) in enumerate(train_loader):
        
        # carry out an optimisation step for each batch
        # after each epoch report the average loss over the training dataset
        
        optimizer.zero_grad()
        recon_batch, mu, logvar = model(data)
        loss = loss_function(recon_batch, data, mu, logvar)
        loss.backward()
        train_loss += loss.item()
        optimizer.step()
        
        if batch_idx % 100 == 0:
            print('Train Epoch: {} [{}/{} ({:.0f}%)]\tLoss: {:.6f}'.format(
                epoch, batch_idx * len(data), len(train_loader.dataset),
                100. * batch_idx / len(train_loader),
                loss.item() / len(data)))
    print('====> Epoch: {} Average loss: {:.4f}'.format(
          epoch, train_loss / len(train_loader.dataset)))


def test(epoch):
    model.eval()
    
    test_loss = 0
    with torch.no_grad():
        for i, (data, _) in enumerate(test_loader):
            
            # report the average loss over the test dataset
            
            recon_batch, mu, logvar = model(data)
            test_loss += loss_function(recon_batch, data, mu, logvar).item()
    test_loss /= len(test_loader.dataset)
    print('====> Test set loss: {:.4f}'.format(test_loss))

for epoch in range(1, 5):
    train(epoch)
    test(epoch)
```

**solution**

For each batch use `optimizer.zero_grad()`, compute `recon, mu, logvar`, evaluate the loss, call `backward()`, and finally `optimizer.step()`. Accumulate total loss across batches and divide by the number of examples. During evaluation call `model.eval()` and use `torch.no_grad()`; the stochastic reparameterisation still means estimated losses can vary slightly between runs.

### Latent space interpolation

Now that we have learnt a latent representation, we can interpolate smoothly between two digits by linearly interpolating between them in the latent space.

**Task 5:** show the mean of the likelihood (image reconstruction) for a set latent vectors which are linearly interpolated between two test images in the latent space.

```python

with torch.no_grad():
    
    mu1 = model(test_dataset[10][0])[1]
    mu2 = model(test_dataset[123][0])[1]
    
    plt.figure(figsize=(20,2))
    
    for i,k in enumerate(np.linspace(0,1,10)):
        
        mu = mu1 + k*(mu2-mu1)
        recon = model.decode(mu)
        
        plt.subplot(1,10,i+1)
        plt.imshow(recon.view(28, 28))
        plt.axis("off")
    plt.show()
```

**solution**

The interpolation uses $z(t)=(1-t)\mu_1+t\mu_2$, for $t\in[0,1]$. Smooth-looking image changes suggest that nearby latent vectors decode to related outputs, though a smooth interpolation is not a proof of a meaningful global latent representation.

**solution — Smaller latent dimension:** Decreasing from 20 to a smaller number restricts information capacity. Reconstructions may worsen, although a compact latent space can sometimes improve structure; compare reconstruction loss, KL and visual quality rather than assuming one direction. To implement, change both encoder latent heads (`fc21`, `fc22`) and decoder input (`fc3`) to the new dimension, then retrain.

**solution — β-VAE:** Use `reconstruction_loss + beta * KLD` with `beta > 1`. Stronger KL pressure generally encourages posterior regularity/disentanglement but may reduce reconstruction fidelity or produce posterior collapse; empirical outcomes depend on β and architecture.

**solution — Binary cross-entropy:** With pixel inputs scaled to `[0,1]` and sigmoid decoder, use `F.binary_cross_entropy(recon, x.view(-1,784), reduction="sum") + KLD`. This corresponds to a Bernoulli-likelihood treatment (often used for binarised MNIST). The change alters the observation model; results may differ from the Gaussian squared-error version.

**solution — Latent-space clustering:** Encode held-out images to posterior means `mu` and cluster those vectors (e.g. k-means); compare clusters with known digit labels using a confusion table or adjusted Rand index. Separate labels from training the unsupervised VAE, and do not claim the learned clusters must perfectly match digits.
